Neste notebook, realizamos uma primeira interação com um **Large Language Model (LLM)** utilizando Python e uma API.

O material original da aula utiliza a **API da OpenAI**. Para esta atividade, a implementação foi adaptada para utilizar a **GroqCloud API**, que possui compatibilidade com o SDK da OpenAI. Dessa forma, foi possível manter a estrutura proposta na aula, alterando apenas a configuração do cliente, a chave de API e o modelo utilizado.


## 1. Importação de Bibliotecas e Configuração

Primeiro, importamos as bibliotecas necessárias, carregamos as variáveis de ambiente definidas no arquivo `.env` e inicializamos o cliente apontando para o endpoint da GroqCloud.

A chave da API permanece fora do código-fonte para evitar a exposição de informações sensíveis.


````python
import os
from dotenv import load_dotenv
from openai import OpenAI

# Carrega as variáveis de ambiente do arquivo .env
load_dotenv()

# Inicializa o cliente utilizando a API da GroqCloud
client = OpenAI(
    api_key=os.getenv("GROQ_API_KEY"),
    base_url="https://api.groq.com/openai/v1"
)
````

## 2. Definição do Modelo

O modelo é carregado através da variável `OPENAI_MODEL` configurada no arquivo `.env`.

Exemplo de configuração:

```env
GROQ_API_KEY=sua_chave_aqui
OPENAI_MODEL=openai/gpt-oss-20b
```

Caso a variável `OPENAI_MODEL` não esteja definida, o código utiliza `openai/gpt-oss-20b` como valor padrão.


````python
# Usa o modelo definido no .env ou um valor padrão
modelo = os.getenv("OPENAI_MODEL", "openai/gpt-oss-20b")

print(f"Modelo utilizado: {modelo}")
````

## 3. Chamada à API

Nesta etapa, enviamos uma mensagem ao modelo utilizando `chat.completions.create()`.

O parâmetro `store=True`, presente no exemplo original com a OpenAI, foi removido nesta versão para manter a compatibilidade com a API da GroqCloud.


````python
response = client.chat.completions.create(
    model=modelo,
    messages=[
        {"role": "user", "content": "write a haiku about ai"}
    ]
)

print(response.choices[0].message.content)
````

## 4. Testando um Novo Prompt

Também podemos alterar o conteúdo enviado ao modelo para testar diferentes perguntas e observar as respostas geradas.


````python
response = client.chat.completions.create(
    model=modelo,
    messages=[
        {"role": "user", "content": "Qual a capital do Brasil?"}
    ]
)

print(response.choices[0].message.content)
````
